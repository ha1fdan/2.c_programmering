```python
from sympy import *
init_printing(use_unicode=False)
t,s = symbols('t s')

#r
def rx(t):
    return -204*t+17
def ry(t):
    return 432*(t**2)-72*t+3

#s
def bx(s):
    return 70*s+2
def by(s):
    return 140*(s**2)

res=solve([Eq(rx(t),bx(s)),Eq(ry(t),by(s))], t,s, dict=True)
for ress in res:
    x,y=ress.values()
    _=res.index(ress)+1
    x_value=round(rx(float(y)),11)
    y_value=round(ry(float(y)),11)
    print(f"P{_}: ({x_value}, {y_value})")
```

```python
import numpy as np
import matplotlib.pyplot as plt

t_vals = np.linspace(-1, 1, 20000)

rx_vals = []
ry_vals = []
for t_val in t_vals:
    rx_vals.append(rx(t_val))
    ry_vals.append(ry(t_val))

bx_vals = []
by_vals = []
for t_val in t_vals:
    bx_vals.append(bx(t_val))
    by_vals.append(by(t_val))
    
plt.figure(figsize=(10, 5))
plt.plot(rx_vals, ry_vals, color='crimson', label='Rødt fly r(t)')
plt.plot(bx_vals, by_vals, color='steelblue', label='Blåt fly r(t)')

for ress in res:
    n=int(res.index(ress))+1
    x,y=ress.values()
    x_value=rx(float(y))
    y_value=ry(float(y))
    plt.scatter(x_value, y_value, zorder=5, s=50, marker='o',
                color='darkorange',
                label=f"P{n} ≈ ({x_value:.2f}, {y_value:.2f})") # plot punkter
    
    plt.annotate(f"P{n} ≈ ({x_value:.2f}, {y_value:.2f})", 
                xy=(x_value, y_value), xytext=(x_value-0.5, y_value+0.15),
                arrowprops=dict(arrowstyle='->', color='darkorange'),
                fontsize=10, color='darkorange') # pil med tekst
    
plt.axhline(y=0, color='black', linewidth=1)   # Landingsbane
plt.title('Opgave 6: To flys banekurver')
plt.xlabel('x-værdi')
plt.ylabel('y-værdi')
plt.xlim(left=0, right=7)
plt.ylim(top=1,bottom=-0.1)
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sympy import symbols, diff, sqrt
t = symbols('t')
rx = -204*t + 17
ry = 432*t**2 - 72*t + 3

dx = diff(rx, t) # rx'(t) = -204
dy = diff(ry, t) # ry'(t) = 864t - 72

fart = sqrt(dx**2 + dy**2)
fart_008 = fart.subs(t, 0.08)

print(f'Fart ved t=0.08: {fart_008:.4f} km/t')
```
