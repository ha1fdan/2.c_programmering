```python
from sympy import *
init_printing(use_unicode=False)
t,s = symbols('t s')

#r
def rx(t):
    return -204*t+17
def ry(t):
    return 432*(t**2)-72*t+3

#s
def bx(s):
    return 70*s+2
def by(s):
    return 140*(s**2)

#x=solve(Eq(rx(t),bx(s)),t)
#y=solve(Eq(ry(x[0]),by(s)),s)

#for y_val in y:
    #val=y.index(y_val)+1
    #print(f"P{val}: ({round(bx(float(y_val)),11)}, {round(by(float(y_val)),11)})")

res=solve([Eq(rx(t),bx(s)),Eq(ry(t),by(s))], t,s, dict=True)
for ress in res:
    x,y=ress.values()
    _=res.index(ress)+1
    x_value=round(rx(float(y)),11)
    y_value=round(ry(float(y)),11)
    print(f"P{_}: ({x_value}, {y_value})")
```
