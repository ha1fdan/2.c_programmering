```python
from sympy import *
init_printing(use_unicode=False)
t,s = symbols('t s')

#r
def rx(t):
    return -204*t+17
def ry(t):
    return 432*(t**2)-72*t+3

#s
def bx(s):
    return 70*s+2
def by(s):
    return 140*(s**2)

x=solve(Eq(rx(t),bx(s)),t)
y=solve(Eq(ry(x[0]),by(s)),s)

solve(Eq(x(t),0),t)

for y_val in y:
    val=y.index(y_val)+1
    print(f"P{val}: ({round(bx(float(y_val)),11)}, {round(by(float(y_val)),11)})")
```

```python
import numpy as np
import matplotlib.pyplot as plt

t_vals = np.linspace(0, 1, 20000) #20000 er precision

rx_vals = []
for t_val in t_vals:
    rx_vals.append(rx(t_val))

ry_vals = []
for t_val in t_vals:
    ry_vals.append(ry(t_val))

plt.figure(figsize=(6, 4))
plt.plot(rx_vals, ry_vals)
plt.title('rx(t)')
plt.xlabel('t')
plt.ylabel('rx')
plt.xlim(left=0, right=20)
plt.ylim(top=5,bottom=-1)
plt.grid(True)
plt.show()
```
