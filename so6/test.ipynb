## Solve for 3x=9

```python
from sympy import *
init_printing() # Latex 2d output
x=symbols('x')
solve(Eq(3*x, 9),x)
```

# Vector

```python
import numpy as np

t=symbols("t")
r_x=Eq(t**2+(1/t),0)
r_y=Eq(3*sqrt(t),0)
y=solve(r_x,t)
x=solve(r_y,t)

f_x = lambdify(t, r_x, "numpy")
f_y = lambdify(t, r_y, "numpy")

# Evaluate over a range of t
t_vals = np.linspace(0.1, 10, 1000)
x_vals = f_x(t_vals)
y_vals = f_y(t_vals)

# Plot
plt.plot(x_vals, y_vals)
plt.xlabel="x"
plt.ylabel="y"
plt.ylim = (-2,8)
plt.show()
```
