## Solve for 3x=9

```python
from sympy import *
init_printing() # Latex 2d output
x=symbols('x')
solve(Eq(3*x, 9),x)
```

```python
import matplotlib.pyplot as plt
plt.plot([1,2,3],[4,5,6], 'r-*')
plt.xlabel = "x"
plt.ylabel = "y"
plt.ylim = (-2,8)
plt.show()
```

# Vector

```python
import numpy as np
from sympy import *
t=symbols("t")
r_x=Eq(t**2,0)
r_y=Eq(3*t,0)
y=solve(r_x,t)
x=solve(r_y,t)

print(x,y)
```

```python
import matplotlib.pyplot as plt
plt.plot(x,y, 'r-*')
plt.xlabel = "x"
plt.ylabel = "y"
plt.ylim = (-2,8)
plt.show()
```
